# Deploy Semantic Model

Simple notebook that uses [semantic-link-labs](https://github.com/microsoft/semantic-link-labs) to deploy a semantic model from one workspace to another via a deployment pipeline.

```python
%pip install semantic-link-labs --quiet
```

```python
from sempy_labs import deployment_pipeline as dp

deployment_pipeline_name = "My Deployment Pipeline"  # name or ID of the deployment pipeline
semantic_model = "My Semantic Model"                 # name of the semantic model to deploy
source_stage = "Development"
target_stage = "Test"

run_deployment = True  # set to False to skip the deployment
```

```python
if run_deployment:
    stages = dp.list_deployment_pipeline_stages(deployment_pipeline=deployment_pipeline_name).set_index("Deployment Pipeline Stage Name")
    items = dp.list_deployment_pipeline_stage_items(deployment_pipeline=deployment_pipeline_name, stage=source_stage)

    dp.deploy_stage_content(
        deployment_pipeline=deployment_pipeline_name,
        source_stage_id=stages.loc[source_stage, "Deployment Pipeline Stage Id"],
        target_stage_id=stages.loc[target_stage, "Deployment Pipeline Stage Id"],
        items={
            "sourceItemId": items.loc[items["Item Display Name"] == semantic_model, "Item Id"].iloc[0],
            "itemType": "SemanticModel",
        },
    )
```
